```python
import os
from dotenv import load_dotenv

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import Tool
from langchain_tavily import TavilySearch

# Load API Keys
load_dotenv(dotenv_path=".env")

# Initialize LLM
llm = ChatOpenAI(
    model="gpt-4.1-mini",
    temperature=0,
    api_key=os.getenv("OPENAI_API_KEY")
)
#  Tool 1: Simple QA

qa_prompt = ChatPromptTemplate.from_template(
    "Answer clearly: {question}"
)

qa_chain = qa_prompt | llm

qa_tool = Tool(
    name="Simple QA",
    func=lambda question: qa_chain.invoke(
        {"question": question}
    ).content,
    description="Answer factual questions clearly"
)

#  Tool 2: Summarizer

summary_prompt = ChatPromptTemplate.from_template(
    "Summarize the following text:\n\n{text}"
)

summary_chain = summary_prompt | llm

summary_tool = Tool(
    name="Summarizer",
    func=lambda text: summary_chain.invoke(
        {"text": text}
    ).content,
    description="Summarizes input text"
)

# Tool 3: Web Search (Tavily)

search_tool = TavilySearch(
    max_results=3,
    tavily_api_key=os.getenv("TAVILY_API_KEY")
)

#  Test Inputs

qa_query = "What is LangGraph in LangChain?"

summary_text = """
LangGraph is a framework for building stateful multi-step agents using LangChain.
It uses graph-based design to model agent workflows and memory.
"""

search_query = "Latest updates on GPT-4o by OpenAI"

# 🚀 Run Tools

print("\n Simple QA Tool Output:\n")
print(qa_tool.invoke(qa_query))

print("\n Summarizer Tool Output:\n")
print(summary_tool.invoke(summary_text))

print("\n Web Search Tool Output:\n")
print(search_tool.invoke(search_query))
```
