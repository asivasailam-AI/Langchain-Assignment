```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate #change 
from langchain_core.tools import Tool

# 🔐 Load API key from .env
load_dotenv(dotenv_path=".env")

# 🔸 Step 1: Initialize the LLM
llm = ChatOpenAI(
    model="gpt-4.1-mini",   # or another supported OpenAI model
    temperature=0,
    api_key=os.getenv("OPENAI_API_KEY")
)

# 🔸 Step 2: Create a Chat Prompt Template
prompt = ChatPromptTemplate.from_template(
    "Answer clearly: {question}"
)

# 🔸 Step 3: Create the LCEL Chain
qa_chain = prompt | llm

# 🔸Step 4: Wrap LCEL Chain as a Tool
qa_tool = Tool(
    name="Simple QA",
    func=lambda question: qa_chain.invoke(
        {"question": question} #Lambda is simply a concise way to adapt the tool's string input to the dictionary input expected by the LCEL chain and to return only the generated text.
    ).content,
    description="A basic LLM chain that answers clear questions"
)

# 🔸 Step 5: Use the Tool
query = "What is LangGraph in LangChain?"

answer = qa_tool.invoke(query)

# 🖨️ Output
print("\nUser Question:", query)
print("\nTool Answer:", answer)
```
